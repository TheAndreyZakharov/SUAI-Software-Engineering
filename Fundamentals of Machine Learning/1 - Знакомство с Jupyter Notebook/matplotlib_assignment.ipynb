## Часть 2. Вычисления в Python

Напишите функцию на Python, которая вычисляет математическую функцию $y(x)$ с заданными скалярными параметрами $a$, $b$, $c$, $d$ (если применимо) и списком значений независимой переменной $x$. Найти математические функции, доступные в библиотеке `numpy`, можно [здесь](https://numpy.org/doc/stable/reference/routines.math.html).

Пример для функции $y(x) = a \sin^2x + b\log_cx$ может выглядеть так:
```python
def my_function(x,a,b,c,d):
    return a * np.sin(x) ** 2 + b * np.log(x) / np.log(c)
```

````python
def my_function(x, a, b, c, d):
    # Вычисление y(x) по формуле ax^2 + by^2 = c^2
    # Решаем уравнение относительно y
    return np.sqrt((c**2 - a * x**2) / b)
````

**Задайте** значения для параметров $a$, $b$, $c$, $d$ и укажите диапазон для переменной $x$:

````python
### BEGIN YOUR CODE

# Присваиваем значения параметрам в виде кортежей
a = 2, 3, 1
b = 1, 2, 0.5
c = 5, 6, 4
d = None

x = np.linspace(0.1, 2, 100)  # Изменяем диапазон, чтобы избежать комплексных значений


### END YOUR CODE
````

Вычислите значения функции $y(x)$ на интервале `x`, векторно и без использования циклов. Постройте график функции. Измените значения параметров $a$, $b$, $c$, $d$ и $x$ в ячейке выше, чтобы получить аккуратный, понятный и красивый график.

````python


# Создаем графики
fig, axes = plt.subplots(3, 1, figsize=(10, 15))  # Изменяем расположение на 3 строки и 1 столбец

# Строим графики для каждого набора параметров
for i in range(3):
    y = my_function(x, a[i], b[i], c[i], d)
    axes[i].plot(x, y, label=f'$a={a[i]}, b={b[i]}, c={c[i]}$', linestyle='--', linewidth=2, marker='o')
    axes[i].set_title(f'$a={a[i]}, b={b[i]}, c={c[i]}$', fontsize=14)
    axes[i].set_xlabel('$x$', fontsize=12)
    axes[i].set_ylabel('$y(x)$', fontsize=12)
    axes[i].grid(True, linestyle=':', color='gray')
    axes[i].legend()

# Общий заголовок
fig.suptitle('Графики функции $y(x) = \\sqrt{\\frac{c^2 - a \\cdot x^2}{b}}$ при разных параметрах', fontsize=16)

# Отображение графиков
plt.tight_layout()
plt.show()


````
